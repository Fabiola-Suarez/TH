# Cuadro comparativo de Máquinas de Turing

| Característica | MT aceptable | MT calculable |
| --- | --- | --- |
| **¿Qué estudia?** | Lenguajes y cadenas | Funciones y resultados |
| **Entrada** | Una cadena `w` | Un valor o conjunto de valores |
| **Objetivo** | Determina si una cadena pertenece a un lenguaje | Obtiene el resultado de una función |
| **Si la entrada corresponde** | La MT se detiene y acepta | La MT se detiene y devuelve el resultado |
| **Si la entrada no corresponde** | Puede rechazar o quedar ejecutándose para siempre | Si la función no está definida para esa entrada, puede no detenerse |
| **Resultado principal** | Aceptar / no aceptar | Un valor calculado |
| **Ejemplo** | `L = {0ⁿ1ⁿ : n ≥ 0}` | `f(n) = n + 1` |
| **Ejemplo de entrada** | `0011` | `5` |
| **Resultado** | Acepta porque tiene dos 0 y dos 1 | Devuelve `6` |
